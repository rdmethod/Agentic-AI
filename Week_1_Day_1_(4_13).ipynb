### What is Python?
- High-level, interpreted programming language
- Created by Guido van Rossum in 1991
- Known for readability and simplicity
- Popular in AI, ML, data science, web development

### Jupyter Notebook vs Google Colab

#### **Jupyter Notebook**
- Runs locally on your machine
- Requires Python installation
- Full control over environment
- No internet needed (after setup)
- Good for: Personal projects, offline work

#### **Google Colab** (Recommended for Beginners!)
- Cloud-based, runs in your browser
- **FREE** GPU and TPU access
- No installation required
- Google Drive integration
- 12-hour session limit (auto-saves work)
- Good for: Learning, experimentation, ML projects

### Troubleshooting Common Issues

**Problem: Can't connect to runtime**
- Solution: Refresh page, try different browser
- Clear browser cache
- Check internet connection

**Problem: Runtime disconnected**
- Solution: Your session expired (12 hours max)
- Click "Reconnect"
- Re-run your cells if needed

**Problem: Out of memory**
- Solution: Runtime → Restart runtime
- Use smaller datasets
- Delete unused variables

**Problem: Package not found**
- Solution: Install it first: `!pip install package-name`
- Restart runtime after installation

**Problem: Code not running**
- Solution: Make sure runtime is connected
- Check for syntax errors (red underlines)
- Read error messages carefully

## Variables and Data Types

### Variables
Variables are containers for storing data values. Python is dynamically typed (no need to declare types).

````python
# Variable assignments
# Variables are like labeled containers that store values
# Python is dynamically typed - no need to declare the type explicitly

name = "Hari"                # String: text data enclosed in quotes
age = 25                      # Integer: whole numbers without decimal points
height = 5.6                  # Float: numbers with decimal points
is_student = True             # Boolean: True or False (note the capital T and F)

# Using f-strings to print formatted output
# f-strings allow you to embed variables directly in strings using {}
print(f"Name: {name}")
print(f"Age: {age}")
print(f"Height: {height} feet")
print(f"Is Student: {is_student}")

# Check types using the type() function
# type() tells you what kind of data a variable holds
print(f"\nType of name: {type(name)}")           # Will show: <class 'str'>
print(f"Type of age: {type(age)}")               # Will show: <class 'int'>
print(f"Type of height: {type(height)}")         # Will show: <class 'float'>
print(f"Type of is_student: {type(is_student)}") # Will show: <class 'bool'>
````

````python
# Implicit Type Conversion Examples
# Python automatically converts types when needed

# Example 1: Integer + Float = Float (automatic conversion)
int_num = 10
float_num = 3.5
result = int_num + float_num
print(f"Integer {int_num} + Float {float_num} = {result}")
print(f"Result type: {type(result)}")
print("Note: Integer was implicitly converted to float\n")

# Example 2: Operations between int and float
a = 5      # int
b = 2.0    # float
print(f"Multiplication: {a} * {b} = {a * b} (type: {type(a * b)})")
print(f"Division: {a} / {b} = {a / b} (type: {type(a / b)})")
print(f"Power: {a} ** {b} = {a ** b} (type: {type(a ** b)})\n")

# Example 3: Boolean in arithmetic (True=1, False=0)
print("Boolean values in arithmetic:")
print(f"True + 5 = {True + 5}")  # True is treated as 1
print(f"False + 5 = {False + 5}")  # False is treated as 0
print(f"True * 10 = {True * 10}")
print(f"Type of (True + 5): {type(True + 5)}\n")
````

````python
# Explicit Type Conversion - Converting to Integer
print("=" * 50)
print("CONVERTING TO INTEGER - int()")
print("=" * 50)

# Float to int (truncates decimal part)
print(f"int(3.14) = {int(3.14)}")
print(f"int(9.99) = {int(9.99)}")
print(f"int(-7.8) = {int(-7.8)}\n")

# String to int (must contain valid integer)
print(f"int('42') = {int('42')}")
print(f"int('-100') = {int('-100')}")
print(f"int('  50  ') = {int('  50  ')}")  # Handles whitespace
````

````python
# Tuple to list
my_tuple = (1, 2, 3, 4)
print(f"tuple to list: {my_tuple} -> {list(my_tuple)}")

# Set to list
my_set = {3, 1, 4, 1, 5}  # Sets remove duplicates
print(f"set to list: {my_set} -> {list(my_set)}")

# String to list (each character becomes an element)
text = "Hello"
print(f"string to list: '{text}' -> {list(text)}")

# Range to list
print(f"range to list: range(5) -> {list(range(5))}")

# Dictionary to list (gets keys by default)
my_dict = {'a': 1, 'b': 2, 'c': 3}
print(f"dict to list (keys): {list(my_dict)}")
print(f"dict values to list: {list(my_dict.values())}")
print(f"dict items to list: {list(my_dict.items())}\n")

print("=" * 50)
print("CONVERTING TO TUPLE - tuple()")
print("=" * 50)

# List to tuple
my_list = [1, 2, 3, 4]
print(f"list to tuple: {my_list} -> {tuple(my_list)}")

# Set to tuple
print(f"set to tuple: {my_set} -> {tuple(my_set)}")

# String to tuple
print(f"string to tuple: '{text}' -> {tuple(text)}\n")

print("=" * 50)
print("CONVERTING TO SET - set()")
print("=" * 50)

# List to set (removes duplicates)
list_with_dupes = [1, 2, 2, 3, 3, 3, 4]
print(f"list to set: {list_with_dupes} -> {set(list_with_dupes)}")
````
