##  Text File Handling

Python can read, write, and append plain text files using the built-in `open()` function.

### File Modes

| Mode | Meaning | File exists? | File missing? |
|------|---------|-------------|---------------|
| `'w'` | Write (overwrite) | Clears content | Creates new file |
| `'r'` | Read | Reads content | Raises `FileNotFoundError` |
| `'a'` | Append | Adds to end | Creates new file |
| `'x'` | Create exclusively | Raises `FileExistsError` | Creates new file |
| `'r+'`| Read + Write | Reads and writes | Raises `FileNotFoundError` |

> **Always use `with open(...) as f:`** — it automatically closes the file even if an error occurs.

```python
with open("students.txt", "w") as f:
  f.write("Name,Age,Spec\n")
  f.write("Alice,20,CS\n")
  f.write("Bob,21,IT\n")
  f.write("Charlie,22,ME\n")


print("file written successfully")
```

```python
with open("students.txt", "r") as f:
  print(f.read())
```

```python
# ──  Writing CSV with DictWriter ──────────────────────────────────────
# DictWriter lets you write rows as dictionaries — no need to track column order.
import csv

students = [
    {'name': 'Alice',   'age': 20, 'score': 92, 'grade': 'A'},
    {'name': 'Bob',     'age': 22, 'score': 78, 'grade': 'B'},
    {'name': 'Charlie', 'age': 21, 'score': 85, 'grade': 'B'},
    {'name': 'Diana',   'age': 23, 'score': 71, 'grade': 'C'},
    {'name': 'Eve',     'age': 20, 'score': 95, 'grade': 'A'},
]

fieldnames = ['name', 'age', 'score', 'grade']  # defines column order

with open('student_grades.csv', 'w', newline='') as f:
    # newline='' prevents extra blank lines on Windows
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()         # writes the column header row
    writer.writerows(students)   # writes all rows at once

print('student_grades.csv written.')
```

## 5 · Object-Oriented Programming (OOP)

OOP organises code around **objects** — bundles of related data (attributes) and behaviour (methods).

| Concept | Plain English | AI Example |
|---------|--------------|------------|
| **Class** | Blueprint / template | `LLMClient`, `Agent`, `Memory` |
| **Object** | One instance built from the blueprint | `my_agent = Agent(...)` |
| **Attribute** | Data the object stores | `model`, `temperature`, `history` |
| **Method** | Action the object can perform | `chat()`, `remember()`, `reset()` |
| **Inheritance** | Child reuses parent's code | `ClaudeClient(LLMClient)` |
| **Encapsulation** | Hide internal details | Private `__api_key` attribute |
| **Polymorphism** | Same method, different behaviour | Every model implements `.generate()` |

```python
# ── 5a  Defining a class ─────────────────────────────────────────────────

class Student:
    """
    Represents a student in the AI course.

    Attributes
    ----------
    name    : str   — student's full name
    age     : int   — student's age
    scores  : list  — list of assessment scores (0-100)
    """

    # Class variable — shared by ALL instances
    course_name = 'Agentic AI Curriculum'

    def __init__(self, name, age):
        # Instance variables — unique to each student object
        self.name   = name          # store the name
        self.age    = age           # store the age
        self.scores = []            # start with an empty scores list

    # ── Instance methods ──────────────────────────────────────────────────
    def add_score(self, score):
        """Add a single assessment score (must be 0–100)."""
        if not 0 <= score <= 100:
            raise ValueError(f'Score must be 0–100, got {score}')
        self.scores.append(score)

    def average(self):
        """Return the average score; 0.0 if no scores yet."""
        return sum(self.scores) / len(self.scores) if self.scores else 0.0

    def letter_grade(self):
        """Assign A/B/C/D/F based on the average."""
        avg = self.average()
        if avg >= 90: return 'A'
        elif avg >= 80: return 'B'
        elif avg >= 70: return 'C'
        elif avg >= 60: return 'D'
        return 'F'

    # ── Special (dunder) methods ──────────────────────────────────────────
    def __str__(self):
        """Human-readable representation — used by print()."""
        return (f'Student({self.name}, age={self.age}, '
                f'avg={self.average():.1f}, grade={self.letter_grade()})')

    def __repr__(self):
        """Unambiguous representation — used in debugging."""
        return f"Student(name={self.name!r}, age={self.age})"


# ── Create instances ──────────────────────────────────────────────────────
alice = Student('Alice', 20)
bob   = Student('Bob', 22)

# Add scores
alice.add_score(88); alice.add_score(92); alice.add_score(95)
bob.add_score(72);   bob.add_score(68);   bob.add_score(74)

print(alice)                       # calls __str__
print(bob)
print(f'\nClass variable: {Student.course_name}')  # accessed on the class itself
print(f'Alice average : {alice.average():.1f}')
print(f'Bob grade     : {bob.letter_grade()}')

```

```python
# ── 5c  Inheritance — reusing and extending behaviour ────────────────────
# A child class inherits everything from the parent, then adds or overrides
# what it needs. Use super() to call the parent's version of a method.

class Animal:
    """Base class — represents any animal."""

    def __init__(self, name, species):
        self.name    = name
        self.species = species

    def speak(self):
        """Generic — child classes should override this."""
        return f'{self.name} makes a sound.'

    def describe(self):
        """Inherited — child classes get this for free."""
        return f'{self.name} is a {self.species}.'


class Dog(Animal):          # Dog inherits from Animal
    """Dog — knows its breed and can fetch."""

    def __init__(self, name, breed):
        super().__init__(name, species='Dog')  # call parent __init__
        self.breed = breed                      # add dog-specific attribute

    def speak(self):                            # override parent method
        return f'{self.name} barks: Woof!'

    def fetch(self, item):                      # new method — Dogs only
        return f'{self.name} fetches the {item}!'


class Cat(Animal):
    """Cat — knows whether it's indoor and can purr."""

    def __init__(self, name, indoor=True):
        super().__init__(name, species='Cat')
        self.indoor = indoor

    def speak(self):
        return f'{self.name} meows: Meow!'

    def purr(self):
        return f'{self.name} purrs contentedly.'


dog = Dog('Rex', breed='Labrador')
cat = Cat('Whiskers', indoor=True)

# Inherited method
print(dog.describe())       # inherited from Animal — no override
print(cat.describe())

# Overridden method
print(dog.speak())          # Dog's own version
print(cat.speak())          # Cat's own version

# Subclass-specific methods
print(dog.fetch('ball'))
print(cat.purr())

# isinstance() — checks class membership through the hierarchy
print(f'\nisDog an Animal? {isinstance(dog, Animal)}')   # True
print(f'isCat an Animal? {isinstance(cat, Animal)}')   # True
print(f'isDog a Cat?     {isinstance(dog, Cat)}')      # False
```
