## 🧠 Part 1 — What Is "Context" in AI?

### The Simple Definition

When you send a message to an AI model like Gemini, the model doesn't just see your latest message — it sees **everything** in the conversation so far. This entire collection of information — your messages, the AI's previous replies, any instructions you gave at the start — is called the **context**.

Think of it this way:

> **Analogy: A Chef and a Recipe Card**
>
> Imagine you hire a chef. Before cooking, you hand them a card that says:
> - *"You are an Italian chef. Only cook Italian food. Always suggest wine pairings."*
> - *"The customer is allergic to nuts."*
> - *"Previous order: pasta carbonara"*
> - *"New request: suggest a dessert"*
>
> The chef reads ALL of this before deciding what to recommend. This card is the **context**.

In AI terms:
- 📋 **System instruction** = The chef's standing orders ("You are an Italian chef")
- 📝 **Conversation history** = What was already discussed ("Previous order: pasta carbonara")
- 💬 **User message** = The new request ("suggest a dessert")
- 🤖 **AI response** = The chef's recommendation

---

### Why Context Engineering Matters

A poorly designed context leads to:
- ❌ AI that forgets what you told it
- ❌ AI that gives wrong or off-topic answers
- ❌ AI that ignores your instructions
- ❌ Expensive API bills (too many unnecessary tokens)

A well-designed context leads to:
- ✅ Consistent, on-topic responses
- ✅ AI that follows your rules reliably
- ✅ Better answer quality
- ✅ Lower costs

**Context engineering is the skill of designing your context to get the best results.**

## 🪟 Part 2 — The Context Window

### What Is a Context Window?

Every AI model can only "see" a limited amount of text at one time. This limit is called the **context window**. It's measured in **tokens** (we'll explain tokens in a moment).

> **Analogy: Working Memory**
>
> A human has a limited "working memory" — the number of things you can hold in your head at once. If someone gives you a 100-item shopping list and asks you to recite it back, you'll forget most of it because it exceeds your working memory.
>
> An AI model's context window is its "working memory." Information outside the window is simply not accessible to the model during that request.

### Context Window Sizes (as of 2025)

| Model | Context Window | What That Means |
|---|---|---|
| Gemini 1.5 Flash | 1,000,000 tokens | ~750,000 words (~750 novels) |
| Gemini 2.5 Flash | 1,048,576 tokens | ~800,000 words |
| Gemini 2.5 Pro | 1,048,576 tokens | ~800,000 words |
| GPT-4o | 128,000 tokens | ~96,000 words |
| Claude 3.5 Sonnet | 200,000 tokens | ~150,000 words |

**Gemini has one of the largest context windows in the world** — it can process an entire novel in a single request!

### What Goes Into the Context Window?

```
┌────────────────────────────────────────────┐
│            CONTEXT WINDOW                   │
│                                            │
│  🔧 System Instruction (your standing rules)│
│  ────────────────────────────────          │
│  👤 User Message 1                         │
│  🤖 AI Response 1                          │
│  👤 User Message 2                         │
│  🤖 AI Response 2                          │
│  ...                                       │
│  👤 Current User Message ← YOU ARE HERE    │
└────────────────────────────────────────────┘
```

As a conversation grows longer, it uses more of the context window. When you hit the limit, older messages get cut off — the AI literally cannot see them anymore.

## 🔢 Part 3 — What Are Tokens?

### The Basics

AI models don't read text the way humans do — they break text into small pieces called **tokens**. A token is roughly:
- ~4 characters in English
- ~¾ of a word (so 100 words ≈ 133 tokens)

But tokenization is more complex — the model has its own vocabulary of known word pieces:

| Text | Approximate Tokens |
|---|---|
| "Hello" | 1 token |
| "Hello world" | 2 tokens |
| "Retrieval-Augmented Generation" | 4 tokens |
| "pneumonoultramicroscopicsilicovolcanoconiosis" | ~10 tokens |
| A full page of text (~500 words) | ~650 tokens |
| A short book (50,000 words) | ~65,000 tokens |

### Why Tokens Matter

1. **Cost**: APIs charge per token. More tokens = more cost.
2. **Speed**: More tokens = slower responses.
3. **Limits**: You can't exceed the context window size.

### Quick Token Estimation Rule
> **Rule of thumb**: 1 token ≈ 4 characters ≈ 0.75 words
> - 1,000 tokens ≈ 750 words ≈ ~3 pages of text
> - 100,000 tokens ≈ 75,000 words ≈ ~300 pages

## 📚 Part 1: The Journey from Traditional Programming to Generative AI

## 🔄 The AI Evolution Timeline

### Stage 1: Traditional Programming (1940s-1980s)
**How it works:** Humans write explicit rules for computers to follow.

**Example:**
```
IF temperature > 30°C THEN
    Display "It's hot!"
ELSE
    Display "It's not hot"
```

**Limitations:**
- ❌ Can only do what you explicitly program
- ❌ Can't handle unexpected situations
- ❌ Requires rules for every possible scenario

**Real-world example:** A calculator - it only does what you tell it to do.

---

### Stage 2: Machine Learning (ML) - 1990s-2010s
**How it works:** Computers **learn patterns from data** instead of following explicit rules.

**Key Idea:** Instead of programming rules, we show the computer examples, and it figures out the patterns!

**Simple Analogy:**
- Traditional Programming: Teaching a child that "cat" = animal with whiskers, 4 legs, says "meow"
- Machine Learning: Showing a child 1000 pictures of cats, and they learn what cats look like

**Example: Email Spam Filter**
- Traditional way: Write rules like "if email contains 'FREE MONEY', mark as spam"
- ML way: Show the system 10,000 emails (labeled spam/not spam), and it learns patterns

**Types of Machine Learning:**

1. **Supervised Learning** (Learning with a teacher)
   - You give the computer **labeled examples** (input + correct answer)
   - Example: Showing pictures of dogs and cats, each labeled correctly
   - Use cases: Image classification, spam detection, price prediction

2. **Unsupervised Learning** (Finding patterns alone)
   - Computer finds patterns in data **without labels**
   - Example: Grouping customers by shopping behavior without telling it what groups to find
   - Use cases: Customer segmentation, anomaly detection

3. **Reinforcement Learning** (Learning by trial and error)
   - Computer learns by **trying things and getting rewards/penalties**
   - Example: How AlphaGo learned to play Go by playing millions of games
   - Use cases: Game playing, robotics, self-driving cars

**ML Successes:**
- ✅ Netflix recommendations
- ✅ Fraud detection in banking
- ✅ Voice recognition (Siri, Alexa)
- ✅ Medical diagnosis assistance

**ML Limitations:**
- ❌ Needs lots of labeled data (expensive!)
- ❌ Good at specific tasks, can't generalize
- ❌ Can't create new content, only classify/predict

---

### Stage 3: Deep Learning (2010s)
**How it works:** Uses artificial **neural networks** inspired by the human brain.

**What are Neural Networks?**
Think of them as layers of decision-makers working together:
- Each "neuron" looks at a small piece of information
- Layers build up from simple patterns to complex understanding
- Like how your brain recognizes faces: first edges, then features, then whole faces

**Analogy:**
Imagine recognizing a face:
- Layer 1: Detects edges and lines
- Layer 2: Detects eyes, nose, mouth
- Layer 3: Combines features into a complete face
- Layer 4: Recognizes whose face it is

**Breakthrough: Deep Learning can learn from raw data!**
- Old ML: Humans had to manually extract "features" (important characteristics)
- Deep Learning: The network learns what features matter automatically

**Deep Learning Achievements:**
- ✅ Image recognition surpassing humans (ImageNet 2015)
- ✅ Real-time language translation
- ✅ Self-driving car vision systems
- ✅ Voice cloning and synthesis
- ✅ Medical image analysis (X-rays, MRIs)

**Still a limitation:**
- ❌ Mostly **discriminative** (classifying/predicting), not **generative** (creating)

---

### Stage 4: Generative AI (2020s) - **WE ARE HERE!** 🎉

**The Big Shift:** From analyzing and predicting → **Creating new content**

**What is Generative AI?**

**Simple Definition:** AI that can **create new, original content** that has never existed before.

**Key Characteristics:**
- 🎨 Creates (not just classifies)
- 🧠 Understands context and meaning
- 💡 Can be creative and novel
- 🔄 Multimodal (works with text, images, audio, video)

**How it differs from previous AI:**

| Traditional ML | Deep Learning | Generative AI |
|----------------|--------------|---------------|
| Classifies data | Understands patterns | **Creates new content** |
| "This is a cat" | "This is a Siamese cat" | **"Generate a picture of a purple cat wearing a hat"** |
| Predictive | Recognitive | **Creative** |

**Real-World Example:**
- **Traditional ML**: This email is spam (Yes/No)
- **Deep Learning**: This email is 95% likely to be spam because of these patterns
- **Generative AI**: Write me a professional email to decline a meeting politely

## 📝 1. Tokenization: Breaking Text into Pieces

### What is Tokenization?

**Simple Definition:** Tokenization is the process of breaking text into smaller units called **tokens** that the AI can understand.

**Why needed?** Computers don't understand words or letters directly - they only understand numbers! Tokenization is the first step in converting human language into something a computer can process.

**Analogy:**
Think of tokenization like breaking a LEGO structure into individual LEGO pieces before you can analyze or rebuild it differently.

---

### How Tokenization Works

**Example 1: Simple Tokenization**

Text: `"Hello, world!"`

Possible tokenization:
- **Word-based**: `["Hello", ",", "world", "!"]` (4 tokens)
- **Character-based**: `["H", "e", "l", "l", "o", ",", " ", "w", "o", "r", "l", "d", "!"]` (13 tokens)
- **Subword-based** (what LLMs use): `["Hello", ",", "world", "!"]` (4 tokens)

---

### Why Subword Tokenization?

Modern LLMs use **subword tokenization** - a smart middle ground:

**Advantages:**
- ✅ Handles unknown words by breaking them into known pieces
- ✅ Efficient vocabulary size (not too big, not too small)
- ✅ Works across multiple languages
- ✅ Captures word roots and patterns

**Example: How "unhappiness" might be tokenized:**
- Instead of treating it as one unknown word
- Break it down: `["un", "happiness"]` or `["un", "happi", "ness"]`
- The model learns: "un" = negative, "happiness" = emotion

---

### Popular Tokenization Methods

**1. Byte Pair Encoding (BPE)**
- Used by GPT models
- Starts with characters, merges common pairs
- Example: "ed" appears often, so it becomes one token

**2. WordPiece**
- Used by BERT, Gemini
- Similar to BPE but uses likelihood scoring
- Example: "##ing" as a suffix token

**3. SentencePiece**
- Used by Llama, T5
- Language-independent
- Works directly with raw text

---

### Visual Example: Tokenization in Action

## 🔢 2. Encoding: Converting Text to Numbers

### What is Encoding?

**Simple Definition:** Encoding is converting tokens into numerical representations (vectors/embeddings) that capture meaning.

**Key Idea:** Words with similar meanings should have similar number patterns!

---

### The Encoding Process

**Step-by-Step:**

1. **Token to ID Mapping**
   - Each token gets a unique number (ID)
   - Example: "cat" → 2435, "dog" → 7854

2. **ID to Embedding**
   - Each ID is converted to a vector (list of numbers)
   - These numbers encode the token's **meaning**
   - Example: "cat" → [0.2, -0.5, 0.8, 0.1, ...] (hundreds of numbers!)

3. **Contextual Encoding**
   - The Transformer processes these vectors
   - Updates them based on surrounding words
   - "bank" near "river" vs "bank" near "money" = different embeddings!

---

### Embeddings: The Magic of Meaning

**What are Embeddings?**
- Numerical representations that capture semantic meaning
- Typically 768, 1024, or more dimensions
- Similar words have similar embeddings

**Analogy:**
Think of embeddings like GPS coordinates:
- "King" might be at coordinates [0.5, 0.8, 0.2, ...]
- "Queen" might be nearby at [0.5, 0.7, 0.3, ...]
- "Cat" would be far away at [-0.2, 0.1, -0.8, ...]

**Famous Example:**
```
king - man + woman ≈ queen
```
This actually works with word embeddings!

---

### Visual Representation

After Encoding:
┌─────┐
│ The │ → [0.12, -0.34, 0.56, ...] (768 numbers)

```

## 🧠 3. The Transformer Architecture: The Engine

### What is a Transformer?

**Simple Definition:** A Transformer is a neural network architecture that processes all tokens simultaneously and learns relationships between them.

**Revolutionary because:**
- ✅ Processes entire sequences in parallel (fast!)
- ✅ Captures long-range dependencies
- ✅ Uses "attention" to focus on relevant parts

---

### Key Components

**1. Self-Attention Mechanism**

**What it does:** Allows each word to "look at" all other words and understand context.

**Example:**
Sentence: "The animal didn't cross the street because **it** was too tired."

**Question:** What does "it" refer to?
- The animal? ✅
- The street? ❌

**How attention helps:**
```
"it" pays attention to:
- "animal" → HIGH attention score (0.8)
- "street" → LOW attention score (0.1)
- "tired" → MEDIUM attention score (0.5)
```

The model learns "it" refers to "animal"!

---

**2. Multi-Head Attention**

**Analogy:** Instead of one pair of eyes, the model has multiple "heads" (like 8 or 12 pairs of eyes) looking at different aspects:
- Head 1: Focuses on grammar
- Head 2: Focuses on semantic relationships
- Head 3: Focuses on sentence structure
- etc.

---

**3. Feed-Forward Networks**

After attention, each token's representation goes through:
- Neural network layers
- Learns complex patterns
- Transforms the embedding

---

**4. Layer Normalization & Residual Connections**

**Purpose:**
- Keeps training stable
- Prevents information loss
- Allows very deep networks (100+ layers!)

---

│  - Multi-Head Attention │
│  - Feed-Forward Network │
└─────────────────────────┘
     ↓
[Repeat 12-96 times depending on model size]
     ↓
[Output: Rich contextual representations]
```

## 🔄 4. Auto-Regression: How LLMs Generate Text

### What is Auto-Regression?

**Simple Definition:** Auto-regression is when the model generates text **one token at a time**, using its own previous outputs as input for the next prediction.

**Key Idea:** Each new word depends on all the words that came before it.

---

### How Auto-Regressive Generation Works

**Step-by-Step Example:**

**Your Prompt:** "The capital of France is"

**Generation Process:**

```
Step 1:
Input:  "The capital of France is"
Output: "Paris" (predicted)

Step 2:
Input:  "The capital of France is Paris" ← includes previous output
Output: "." (predicted)

Step 3:
Input:  "The capital of France is Paris."
Output: "It" (predicted)

Step 4:
Input:  "The capital of France is Paris. It"
Output: "is" (predicted)

Step 5:
Input:  "The capital of France is Paris. It is"
Output: "known" (predicted)

... continues until stopping condition
```

**Final Output:** "The capital of France is Paris. It is known for its art, culture, and cuisine."

---

### The Auto-Regressive Loop

```
┌─────────────────────────────────────┐
│                                     │
│   1. Start with prompt              │
│   2. Encode to embeddings           │
│   3. Pass through Transformer       │
│   4. Predict next token             │
│   5. Add predicted token to input   │ ←─┐
│   6. Repeat from step 2             │   │
│                                     │   │
└─────────────────────────────────────┘   │
          │                               │
          └───────────────────────────────┘
                (Loop continues)
```

---

### Why "Auto-Regressive"?

- **Auto:** Self (the model uses its own outputs)
- **Regressive:** Predicts based on previous values

**Mathematical View:**
```
P(word₁, word₂, word₃, ..., wordₙ) =
P(word₁) ×
P(word₂ | word₁) ×
P(word₃ | word₁, word₂) ×
P(word₄ | word₁, word₂, word₃) ×
...
```

Each word is predicted based on ALL previous words!

---

### Controlling Generation

**1. Temperature** (covered earlier)
- Controls randomness
- Low = predictable, High = creative

**2. Top-k Sampling**
- Only consider the k most likely next tokens
- Example: k=5 means pick from top 5 predictions

**3. Top-p (Nucleus) Sampling**
- Consider tokens until cumulative probability reaches p
- More dynamic than top-k

**4. Beam Search**
- Explore multiple possible sequences
- Choose the best overall sequence

## 🔓 5. Decoding: Converting Numbers Back to Text

### What is Decoding?

**Simple Definition:** Decoding is the process of converting the model's numerical predictions back into human-readable text.

**The Final Step:** Numbers → Tokens → Text

---

### The Decoding Process

**Step-by-Step:**

```
1. Model Output (Logits)
   [2.3, -1.5, 4.7, 0.8, ...] ← One number for EACH token in vocabulary
   ↓
2. Apply Softmax (Convert to Probabilities)
   [0.02, 0.001, 0.89, 0.04, ...]
   ↓
3. Sample/Select Next Token
   Token ID: 3425 (based on probability distribution)
   ↓
4. Convert Token ID to Token
   3425 → "Paris"
   ↓
5. Detokenize (if needed)
   Combine subword tokens: ["un", "##happy"] → "unhappy"
   ↓
6. Return Text to User
   "Paris"
```

---

### Greedy vs. Sampling Decoding

**Greedy Decoding:**
- Always pick the highest probability token
- Fast but can be repetitive
- Example: Always picks "the" as most likely article

**Sampling Decoding:**
- Pick tokens based on probability distribution
- More diverse outputs
- Controlled by temperature

**Example:**

```
Prompt: "The weather is"

Probability Distribution:
- "sunny" → 40%
- "nice" → 25%
- "beautiful" → 20%
- "pleasant" → 10%
- "good" → 5%

Greedy Decoding → Always picks "sunny"

Sampling Decoding →
- 40% chance of "sunny"
- 25% chance of "nice"
- 20% chance of "beautiful"
- etc.
```

## 🎯 7. Key Concepts Summary

### Tokenization
- **What:** Breaking text into processable units
- **How:** Subword algorithms (BPE, WordPiece, SentencePiece)
- **Why:** Computers need discrete units to process

### Encoding
- **What:** Converting tokens to numerical vectors
- **How:** Embedding layers + Transformer layers
- **Why:** Capture meaning in mathematical form

### Transformer
- **What:** Neural network architecture
- **How:** Self-attention + Feed-forward layers
- **Why:** Understand context and relationships

### Auto-Regression
- **What:** Sequential token-by-token generation
- **How:** Predict next token, add to input, repeat
- **Why:** Generate coherent, contextual text

### Decoding
- **What:** Converting predictions back to text
- **How:** Logits → Probabilities → Token selection
- **Why:** Make output human-readable
